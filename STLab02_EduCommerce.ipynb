## 2. Chạy 33 test cases


```python
import subprocess
command = [sys.executable, 'test_runner.py', '--cases', 'test_cases.csv', '--out', 'reports']
completed = subprocess.run(command, text=True, capture_output=True)
print(completed.stdout)
if completed.stderr:
    print('STDERR:', completed.stderr)
print('Runner exit code:', completed.returncode)
```

## 3. Xem report và kiểm tra số case đã nạp


```python
import pandas as pd
from IPython.display import display
summary = pd.read_csv('reports/test_summary.csv')
execution = pd.read_csv('reports/test_execution.csv', keep_default_na=False)
display(summary)
display(execution[['test_case_id','requirement_id','execution_status','execution_time_ms','error_message']])
print('Loaded:', len(execution), ' / PASS+FAIL+BLOCKED:', execution['execution_status'].isin(['PASS','FAIL','BLOCKED']).sum())
```
